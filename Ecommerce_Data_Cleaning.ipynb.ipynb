```python
# Import Required Libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
# Explore the Dataset
df.head()
```

```python
# View  Dataset Information
df.info()
```

```python
# Statistical Summary
df.describe()
```

```python
# Dataset Shape
df.shape
```

```python
# Check Missing Values
df.isnull().sum()
```

```python
# Check Duplicate Records

df.duplicated().sum()
```

```python
# Remove Duplicatte Records

df.drop_duplicates(inplace = True)
```

```python
# check any Duplicate value
df.duplicated().sum()
```

```python
# Replace Invalid Values
# Replace common invalid values(N/A, Null, space) with NaN

df.replace(['N/A','Null',''],np.nan,inplace = True)
```

```python
# Remove Extra Spaces 

df = df.apply(lambda x: x.str.strip() if x.dtype=='str' else x)
```

```python
# Standardize Text Format
# Convert text into Proper Case.

df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
# Validate gEmail Addresses
# Keep only email records

df = df[df['Email'].str.contains('@',na=False)]
```

```python
# Convert date Columns

df['Order_Date'] = pd.to_datetime(df['Order_Date'],errors='coerce', dayfirst=True)  # jo date ma na wo isko likhne se coerce covert into NAT

df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'],errors='coerce', dayfirst=True)
```

```python
# Convert Numeric Columns

cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
# Remove Invalid Quantity
# Remove records having negative or zero quantity

df = df[df['Qty']>0]
```

```python
# Fill Missing Values

df['Discount'] = df['Discount'].fillna(0)

df['Phone'] = df['Phone'].fillna('Unknown')

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
df.head()
```

```python
# Total Sale by using Create new Column

df['Sales'] = df['Qty']*df['Unit_Price']
```

```python
df.head()
```

```python
# Net Amount by using New column creates

df['Net_Amount']= df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
# Profit column(Example : 20%)
        
df['Profit'] = df['Net_Amount']*0.20
```

```python
# Order Month Column Create

df['Month'] = df['Order_Date'].dt.month_name()
```

```python
# order year column Create

df['Year'] = df['Order_Date'].dt.year
```

```python
# Weekday column Create

df['Weekday'] = df['Order_Date'].dt.day_name()
```
